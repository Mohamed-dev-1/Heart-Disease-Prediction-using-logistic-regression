## Logistic Regression

**Definition :**


* Logistic Regression is a supervised machine learning algorithm used for classification problems. 

    - **Linear regression predicts a numerical value**. 

    - **Logistic regression predicts a probability, then that predicted probability is used to classify**.

  For example in our Hear-disease project :

   - our input is : The person's health features (age , smoking ...etc)

   - We train our Logistic regression model to predicts the probability of is that person belonds to class 1(CHD within 10 years) or class 0 (No CHD within 10 years) 

* Example :

we found probability = 0.80 :

means : estimates an 80% probability that this person belongs to class 1 (CHD within 10 years).


* Logistic Regression is commonly used when there are two possible classes: `Yes / No` , `True / False` , `0 / 1`

* Logistic regression uses the **`Sigmoid`** function to convert inputs(features) into a probability value (the probability value is between 0 and 1).

**Types of Logistic Regression :**

* 1- Binomial Logistic Regression: 
    
    - Used when the dependent variable (values of the target column) **has only two possible categories** like: Yes/No, Pass/Fail, or 0/1

* 2- Multinomial Logistic Regression:

    -  Used when the dependent variable (values of the target column)**has three or more unordered categories** , ex : Red/Blue/Green

* 3- Ordinal Logistic Regression:

    - Used when the dependent variable (values of the target column) **has three or more categories with a natural order** , ex : Low,Medium,High


For our Heart Disease Project we're going to use **Binomial Logistic Regression** , because our target column is `TenYearCHD` and it has only 2 possible categories `0 or 1`

**Assumptions (Data conditions that should be satisfied before Logistic Regression training) :**

* 1- **Independent Observations (data) :**

    - Each row of the data should be **Independent** to the other rows , ex : Row 1's data should not be a direct copy of Row 2's data

* 2- **Binary Dpendent variable :**

    - The values of our dependent variable (the target column `TenYearCHD`) must have only 2 classes, which are: class 0 (No CHD) and class 1 (CHD)

* 3- **Linearity of log-odds :**

    - The independent variables (The features) should have a linear relationship with the log-odds of the dependent variable (values of the target column `TenYearCHD`).

    - What are the log-odds ?

        - **log-odds** is simply : **Log + odds**
        
        - **1- Odds ?** For a binary outcome the result is a probability `p` , Ex : p = 0.8 (means 80% probability of something)

        - The **odds** are : **odds = p / (1 - p)** ,
        
        if p = 0.8 => odds = 0.8 / 0.2
        
        then odds = 4

        **Meaning of odds = 4** : The outcome is 4 times as likely to happen as not happen.

        - **2- Log-odds ?** is simply `Log + p / (1 - p)` , then **it is the Logarithm of odds** : 

            log-odds = **log(  p/(1 - p)  )**


    - So , the chain is: **Probability** then **Odds** then **Log-Odds**


* 4- **No Extreme Outliers :**

    - Extreme outliers :values that are far away from most of the other values , ex: 10, 20, 30, **200** (200 is an extreme value)

    -  The probelem Extreme outliers cause :

        - Logistic Regression is trying to learn the relationship between the features and the target.
        - The extreme value can have too much influence on what the model learns, causing it to calculate other innappropriate weights that will negatively affect model performance. 



* 5- **Large Sample Size :**

    - A sufficiently large dataset helps produce more reliable and stable predictions.

<h3><b>Understanding Sigmoid Function</b></h3>


**The big picture of the Sigmoid function :**

* **Sigmoid is a mathematical function**.

* **converts any real-valued number into a value between 0 and 1**.

* One of its characteristics is the **"S"-shaped curve which is useful in scenarios where we need to convert outputs into probabilities**. 

* This function is **also called the logistic function**.

* **Mathematically**, sigmoid is represented as:

    **σ(x) = 1 / (1 + e⁻ˣ)**

    - Where:
        - **`x`** is **the input value**
        - **`e`** is **Euler's number (≈2.718)** 

**Sigmoid function** is **used as an activation function in machine learning and neural networks for modeling binary classification problems**, smoothing outputs, and introducing non-linearity into models.

---



1. **The sigmoid function is a key component of Logistic Regression** 


2.  The sigmoid function **converts the model's score output (the result of z = w₁x₁ + w₂x₂ + ... + b) into a probability value between 0 and 1** , here is **the flow** :

    - **Features (X)** ,then
            
    - **z** = w₁x₁ + w₂x₂ + ... + b , then
            
    - **Model's raw score** ,then
            
    - **Sigmoid function** , then
            
    - **Probability between 0 and 1**


3. The Sigmoid function **accepts any real number** (means that the model's score value (Z) can be -5, 0, 3, 100) **as an input** , **then converts that input value into a value between 0 and 1**.

    * How doest the Sigmoid function converts any real number into a value between 0 and 1 ?

        -  Simply, Because **the sigmoid formula `σ(x) = 1 / (1 + e⁻ˣ)` is designed to always produce a value between 0 and 1**.
