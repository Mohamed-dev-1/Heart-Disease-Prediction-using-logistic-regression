## Logistic Regression

**Definition :**


* Logistic Regression is a supervised machine learning algorithm used for classification problems. 

    - **Linear regression predicts a numerical value**. 

    - **Logistic regression predicts a probability, then that predicted probability is used to classify**.

  For example in our Hear-disease project :

   - our input is : The person's health features (age , smoking ...etc)

   - We train our Logistic regression model to predicts the probability of is that person belonds to class 1(CHD within 10 years) or class 0 (No CHD within 10 years) 

* Example :

we found probability = 0.80 :

means : estimates an 80% probability that this person belongs to class 1 (CHD within 10 years).


* Logistic Regression is commonly used when there are two possible classes: `Yes / No` , `True / False` , `0 / 1`

* Logistic regression uses the **`Sigmoid`** function to convert inputs(features) into a probability value (the probability value is between 0 and 1).

**Types of Logistic Regression :**

* 1- Binomial Logistic Regression: 
    
    - Used when the dependent variable (values of the target column) **has only two possible categories** like: Yes/No, Pass/Fail, or 0/1

* 2- Multinomial Logistic Regression:

    -  Used when the dependent variable (values of the target column)**has three or more unordered categories** , ex : Red/Blue/Green

* 3- Ordinal Logistic Regression:

    - Used when the dependent variable (values of the target column) **has three or more categories with a natural order** , ex : Low,Medium,High


For our Heart Disease Project we're going to use **Binomial Logistic Regression** , because our target column is `TenYearCHD` and it has only 2 possible categories `0 or 1`

**Assumptions (Data conditions that should be satisfied before Logistic Regression training) :**

* 1- **Independent Observations (data) :**

    - Each row of the data should be **Independent** to the other rows , ex : Row 1's data should not be a direct copy of Row 2's data

* 2- **Binary Dpendent variable :**

    - The values of our dependent variable (the target column `TenYearCHD`) must have only 2 classes, which are: class 0 (No CHD) and class 1 (CHD)

* 3- **Linearity of log-odds :**

    - The independent variables (The features) should have a linear relationship with the log-odds of the dependent variable (values of the target column `TenYearCHD`).

    - What are the log-odds ?

        - **log-odds** is simply : **Log + odds**
        
        - **1- Odds ?** For a binary outcome the result is a probability `p` , Ex : p = 0.8 (means 80% probability of something)

        - The **odds** are : **odds = p / (1 - p)** ,
        
        if p = 0.8 => odds = 0.8 / 0.2
        
        then odds = 4

        **Meaning of odds = 4** : The outcome is 4 times as likely to happen as not happen.

        - **2- Log-odds ?** is simply `Log + p / (1 - p)` , then **it is the Logarithm of odds** : 

            log-odds = **log(  p/(1 - p)  )**


    - So , the chain is: **Probability** then **Odds** then **Log-Odds**


* 4- **No Extreme Outliers :**

    - Extreme outliers :values that are far away from most of the other values , ex: 10, 20, 30, **200** (200 is an extreme value)

    -  The probelem Extreme outliers cause :

        - Logistic Regression is trying to learn the relationship between the features and the target.
        - The extreme value can have too much influence on what the model learns, causing it to calculate other innappropriate weights that will negatively affect model performance. 



* 5- **Large Sample Size :**

    - A sufficiently large dataset helps produce more reliable and stable predictions.

<h3><b>Understanding Sigmoid Function</b></h3>


**The big picture of the Sigmoid function :**

* **Sigmoid is a mathematical function**.

* **converts any real-valued number into a value between 0 and 1**.

* One of its characteristics is the **"S"-shaped curve which is useful in scenarios where we need to convert outputs into probabilities**. 

* This function is **also called the logistic function**.

* **Mathematically**, sigmoid is represented as:

    **σ(x) = 1 / (1 + e⁻ˣ)**

    - Where:
        - **`x`** is **the input value**
        - **`e`** is **Euler's number (≈2.718)** 

**Sigmoid function** is **used as an activation function in machine learning and neural networks for modeling binary classification problems**, smoothing outputs, and introducing non-linearity into models.

---



1. **The sigmoid function is a key component of Logistic Regression** 


2.  The sigmoid function **converts the model's score output (the result of z = w₁x₁ + w₂x₂ + ... + b) into a probability value between 0 and 1** , here is **the flow** :

    - **Features (X)** ,then
            
    - **z** = w₁x₁ + w₂x₂ + ... + b , then
            
    - **Model's raw score** ,then
            
    - **Sigmoid function** , then
            
    - **Probability between 0 and 1**


3. The Sigmoid function **accepts any real number** (means that the model's score value (Z) can be -5, 0, 3, 100) **as an input** , **then converts that input value into a value between 0 and 1**.

    * How doest the Sigmoid function converts any real number into a value between 0 and 1 ?

        -  Simply, Because **the sigmoid formula `σ(x) = 1 / (1 + e⁻ˣ)` is designed to always produce a value between 0 and 1**.

4. **Why is `Sigmoid function` has an S-shaped curve ?**

    * Because if we plot the sigmoid's `x` and `y` values, the drawn graph forms an S shape.

    * Now , let's understand why the sigmoid values form an S-shape graph and what are the values of the sigmoid function ?

    - The sigmoid function formula is : `σ(x) = 1 / (1 + e⁻ˣ)`

    - **The X-axis ?** 
        
        - **contains the model's score values **(The Zs)(each row of the dataset -> has its own model score value *Z*)

    - **The Y-axis ?**

        - for each x value , we calculate the sigmoid `y = σ(x) = 1 / (1 + e⁻ˣ) ` , So : 

        - The Y-axis contains the sigmoid results of each data set row

    - Then we got our values of the X-axis and the Y-axis , so we can plot the coordinates (x,y) , then we got the S-shape graph


    - We got that S-shape curve Because of the mathematical shape of the sigmoid formula : 

       -  The sigmoid creates an S-shape because its output changes slowly near 0 and 1, but changes quickly around the middle (0.5)

5. In logistic regression, **we use a threshold value usually 0.5 to decide the class label**.

    - Classification rule:

        - **Probability ≥ 0.5  ->  Class 1**
        - **Probability < 0.5  ->  Class 0**

    So , The sigmoid gives us a probability, then using the threshold (0.5) we convert that probability into a class label (0 or 1)

    - The process :

        - Features and target values

        - calculate Z score

        - calculate Sigmoid values (The result is a probability value between 0 and 1)

        - use the threshold (0.5) to classify to class 1 or class 0


6. **How "`this probability belongs to class 1 or 0`" is done ?**

    - We calculated `Z` then we put `Z` into the sigmoid function `σ(x) = 1 / (1 + e⁻ˣ)` , which gives us a probability value **σ(x)**.

    - Then , the law says that  : **`P(y=1) = σ(z)`** , means that :

        - **The sigmoid output σ(z) is the probability that \(y\) is 1**, which means : 
        
           - The sigmoid output σ(z) is the probability that (y) **TenYearCHD** is **1** , ex : `P(TenYearCHD = 1) = 0.8`
    
    - The law also says : **`P(y=0)= 1 − σ(z)`** , but where it comes from ?

        - we know that there is only 2 possible classes : class 0 and class 1

        - So their probabilities must add up to 1: **`P(y=0) + P(y=1) = 1`**

        - and we already know that : **`P(y=1) = σ(z)`**

        - So , **`P(y=0) = 1 − σ(z)`**

    - Example :

        - suppose we have : **σ(z) = 0.8**

        - Then : P(TenYearCHD = 1) = 0.8 , means the probability that the person will get CHD within ten years is 80%

        - And P(TenYearCHD = 0) = 1 - 0.8 => P(TenYearCHD = 0) = 0.2 , means the probability that the person will not get CHD within ten years is 20%

        So : 
           - Class 1 -> 80%
           - Class 0 -> 20%

    - The important thing to understand is :

        - **Sigmoid directly gives us the probability of Class 1 . The probability of Class 0 is simply whatever probability is left over** : 
        
        - **[`probability of class 0 = 1 - probability of class 1`]**

    - Question : **Why TenYearCHD = 1 is defined as class 1 and no TenYearCHD = 0 ?**

        - In our dataset:

            - **TenYearCHD = 0**    (No CHD within 10 years)
            - **TenYearCHD = 1**    (CHD within 10 years)

                - **`TenYearCHD = 1` has the value `1` , then Logistic regression treats 'TenYearCHD = 1' as the positive class then , it will be class 1**

## How does `LogisticRegression.fit(X_train, y_train)` work?

Now, let's understand what happens inside **`LogisticRegression.fit(X_train, y_train)`** which tries to **find the weights `w` and bias `b`**.

- Once `fit()` has found those parameters (`w` and `b`), we can use them to calculate the model's raw score **`z`**:

$$
\boxed{z=w\cdot X+b}
$$

So:

- **`fit()` : finds `w` and `b`**

- **`w` and `b` + patient features `X` => calculate `z`**

- **`z` into sigmoid -> gives -> probability**

- **probability compared to threshold (0.5) -> gives the predicted class**


The process :

- fit()
  
- find w and b
  
- z = w·X + b
  
- sigmoid(z)
  
- P(CHD = 1) compared to 0.5 , if P(CHD = 1) >= 0.5 , the value of P(CHD = 1) belongs to class 1 , and the remaining possibility belongs to class 0
  
- 0 or 1


## LogisticRegression.fit() Function's Workflow : 

Now , let's find out **How does `fit()` find the parameters `w` and `b` that will later allow us to calculate `z` ?**

<h3> <b> Step 1 : Start with random values </b> </h3>

* We can think of the model having some initial random  values for **w** and **b** (usually zeros or small random numbers)

<h3> <b> Step 2 : Calculate Z </b> </h3>

* For evey patient (row) , the model calculates : **z = w1​x1 ​+ w2​x2 ​+ ⋯ + wn​xn ​+ b**

* **Z** is the raw score

<h3> <b> Step 3 : Turn Z into a probability </b> </h3>

* Now , we apply the Sigmoid :  **p = `σ(z) = 1 / (1 + e⁻ᶻ)`**

* Suppose **z = 1.5** , then : **p = σ(1.5) ≈ 0.818**

* And because our Class 1 is : TenYearCHD = 1 , we can interpret that as : **P(CHD = 1) = 0.818**

* So , The model currently thinks this patient has an 81.8% probability of belonging to Class 1 (getting CHD within 10 years)

<h3> <b> Step 4 : Measure how wrong the predictions are (using Loss) </b> </h3>

* For each patient (row) we Compare the prediction **p** with the actual value **y**

* The comparison is done using **log loss (binary cross-entropy)** , the formula is :

* **Loss = - [ y * log(p)  +  (1 - y) * log(1 - p) ]**

* Then we Calculate **the average Loss** , denoted: **J(w,b)** 

    - Called **J(w,b)** because the loss depends on the model's w and b, because to calculate the Loss we need p and p = σ(z) , and Z depends on w and b

    - How ? **We calculate the loss for every patient (row), then take the average of all those losses.** , ex : (we have 3 patints and we calculated their Losses) 
    
       -  J(w,b) = Loss1 + Loss2 + Loss3 / 3

    - So : Cost function **J(w,b)** = sum of all Loss values of all patients (rows) / number of all patients rows

* And **Remember** that the The optimizer's goal is :

    - **Find values** of **w** and **b** that make **J(w,b) "The average Loss"** as **small as possible**.

<h3> <b> Step 5 : Calculate the gradient </b> </h3>

Now, the model asks : **My current `w` and `b` are not perfect . In which direction should i change them to make the loss smaller ?**

Using these Formulas :

-  **dw/dJ = 1/m * ​X^T * (p−y)**

and

- **dj/dw = 1/m * ∑ (p−y)**

where : 

   -  `m` = number of training examples (number of patients rows)
   - `X^T` = Transpose of the features matrix


* So , **The gradient tells `fit()` which direction to put `w` and `b` to reduce the loss.**

<h3> <b> Step 6 : Adjust "w" and "b" (using Gradient Descent)</b> </h3>

* If we're using Gradient Descent , then we'd update the weights **w** and the bias **b** using these formulas :

    - **w = w - α * dJ/dw**

    - **b = b - α * dJ/db**

where : 

- `α` is the learning rate (controls step size)

<h3> <b> Step 7 : Repeat</b> </h3>

- Remember **the goal is to find the minimum possible Average Loss J(w,b)**

Soi the model keeps repeating the process until further changes are very small or another stopping condition is reached, like : 

- J = 0.80 , ↓
 
- J = 0.50 , ↓
 
- J = 0.30 , ↓
 
- J = 0.22 , ↓
 
- J = 0.20 , ↓
 
- J = 0.199 , ↓
 
- J = 0.1990    <- At this point barely improving , there is not much benefit from continuing repeating , so the model can stop.

<h3> <b> Step 8 : Store the learned parameters</b> </h3>

* Once **`model.fit(X_train, y_train)`** finishes, the final w and b are stored inside the model object

* `model.coef_` might contain: **[w₁, w₂, w₃, ..., wₙ]**

* `model.intercept_` contains: **b**

Now the model can use them on a new patient (unseen data)
